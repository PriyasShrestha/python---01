```python
import random

class Bank:
    def __init__(self, ac_name, ac_balance):
        self.ac_name = ac_name
        self.ac_balance = ac_balance
        self.ac_number = "".join(str(random.randint(0, 9)) for i in range(16))
        print('Account created')
        print(f'Account Name: {self.ac_name}')
        print(f'Account Number: {self.ac_number}')
        print(f'Initial Balance: Rs.{self.ac_balance}')

    def deposit(self, amount_deposit):
        if amount_deposit >= 500:
            self.ac_balance += amount_deposit
            print(f'Dear {self.ac_name}, Rs.{amount_deposit} has been deposited.')
            print(f'Updated Balance: Rs.{self.ac_balance}')
        else:
            print("Minimum deposit amount is Rs.500")

    def withdraw(self, amount_withdraw):
        if amount_withdraw <= self.ac_balance:
            self.ac_balance -= amount_withdraw
            print(f'Rs.{amount_withdraw} withdrawn.')
            print(f'Remaining Balance: Rs.{self.ac_balance}')
        else:
            print('Insufficient balance...!')

    def show_details(self):
        print(f'Account Name: {self.ac_name}')
        print(f'Account Number: {self.ac_number}')
        print(f'Balance: Rs.{self.ac_balance}')
```
