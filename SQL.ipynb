```python
try:
    cursor.execute('''
        create table Person(
            customer_id INTEGER PRIMARY KEY AUTOINCREMENT,
            first_name TEXT NOT NULL,
            last_name TEXT NOT NULL,
            phone TEXT NOT NULL,
            email TEXT NOT NULL,
            street TEXT NOT NULL,
            city TEXT
        )
    ''')
    conn.commit()
    print("Customer table created")
except sqlite3.OperationalError as e:
    print("Error:", e)
```
