```python
class Mobilebanking(ABC):
    @abstractmethod
    def statement(self):
         pass
    @abstractmethod
    def user_account_details(self):
        pass
    def username(self, name):
        return f'Hello {name}'
    def connection(self):
        return 'The connection have been establish'
        
```

```python
class show_details(Mobilebanking):
    def choose_statement(self):
        return 'choose statement'
     
    def statement(self):
        return 'statement from date 1 to date 3'
   
    def user_account_details(self):
        return 'user account'
```
