## Inheritance

```python
# # multiple classes
# Types of inheritance
# ----------------------
# 1. Simple Inheritance
# 2. Mjultiple inheritance
# 3. Multi-Level Inheritance
'''
Parents and child
child class Inherits/ extends parents class
one school has many students
'''
```

```python
#parents class
class School:
    def __init__ (self, school_name, school_location, school_contact):
        self.school_name=school_name
        self.school_location=school_location
        self.school_contact=school_contact

    def school_details(self):
        print(f'Name of School: {self.school_name}')
        print(f'Location of School: {self.school_location}')
        print(f'Contact of School: {self.school_contact}')
         
        
```

```python
#child class
class Student(School):#class students inherit students class
    def __init__(self, name, section, grade, contact,school_name, school_location, school_contact):
        #super()-> super keyword is used to inherit parent class constructer
        super().__init__(school_name, school_location, school_contact)
        self.name=name
        self.section=section
        self.grade=grade
        self.contact=contact

    def student_details(self):
        print(f'Name of student is : {self.name}')
        print(f'Section of student is : {self.section}')
        print(f'Grade of student is : {self.grade}')
        print(f'Contact of student is :{self.contact}')
        
```

```python
s=Student('Ram','A','5','9867998500','AC school','Bhaktapur','026500525')
s.student_details()
s.school_details()
```

```python
 class Teacher(School):
     def __init__ (self, tname, tcontact, total_classes, school_name, school_location, school_contact):
         super(). __init__ (school_name, school_location, school_contact)
         self.tname=tname
         self.tcontact=tcontact
         self.total_classes=total_classes

     def teacher_details(self):
        print(f'Name of Teacher is : {self.tname}')
        print(f'Contact of Teacher is : {self.tcontact}')
        print(f'Total classes of Teacher is : {self.total_classes}')
       
```

```python
t=Teacher('shyam',986542154,4,'asdf','lkt',26555)
t.teacher_details()
t.school_details()
```
