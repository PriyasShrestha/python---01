```python
class Student:
    def __init__ (self, name, section, no_of_subject):
        self.name = name
        self.section = section
        self.no_of_subject = no_of_subject
        self.school_name = "ABC school"
    def details(self):
        print(f'Name of school is: {self.school_name}')
        print(f'Name of student is: {self.name}')
        print(f'Section of student is: {self.section}')
        print(f'NO of subject of student is: {self.no_of_subject}')
    
```

```python
s=Student('Ram','A','7')
s.details()
```

```python
class Employee:
    def __init__(self, name, designation, salary):
        self.company_name="Tech corporate"
        self.company_department="Development"
        self.name = name
        self.designation = designation
        self.salary = salary
    def display_info(self):
        print(f'Name of Company : {self.company_name}')
        print(f'Name of Department: {self.company_department}')
        print(f'Name of Employee : {self.name}')
        print(f'Designation : {self.designation}')
        print(f'Salary of employee: {self.salary}')
        
```

```python
s=Employee('Ram','JR officer','25000')
s.display_info()
```

```python
class Developer:
    def __init__ (self, name, primary_language, years_of_experience):
        self.role="Software Enginner"
        self.enviroment="Remote"
        self.name=name
        self.primary_language=primary_language
        self.years_of_experience=years_of_experience
    def coding_status(self):
        return f'{self.name} is currently coding in {self.primary_language}'
    def experience_level(self):
        if(self.years_of_experience>4):
            return "senior"
        else:
            return"junior"
    
    
    
```
