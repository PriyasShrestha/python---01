## multi level inheritance

```python
class School:
    def __init__ (self, sname, scontact, slocation):
        self.sname = sname
        self.scontact = scontact
        self.slocation = slocation
         
    def school_details(self):
        print(f'Name of school is: {self.sname}')
        print(f'Contact of school is: {self.scontact}')
        print(f'Location of school is: {self.slocation}')
       
```

```python
class Teacher(School):
    def __init__ (self, tname, tsection, tcontact,sname, scontact, slocation):
        super(). __init__(sname, scontact, slocation)
        self.tname = tname
        self.tsection = tsection
        self.tcontact = tcontact
      
    def teacher_details(self):
        print(f'Name of Teacher is: {self.tname}')
        print(f'Section of Teacher is: {self.tsection}')
        print(f'Contact of Teacher is: {self.tcontact}')
```

```python
class Student(Teacher):
    def __init__ (self, name, section, no_of_subject,tname, tsection, tcontact,sname, scontact, slocation):
        super(). __init__(tname, tsection, tcontact,sname, scontact, slocation)
        self.name = name
        self.section = section
        self.no_of_subject = no_of_subject
       
    def student_details(self):
       
        print(f'Name of student is: {self.name}')
        print(f'Section of student is: {self.section}')
        print(f'NO of subject of student is: {self.no_of_subject}')
```

```python
s=Student('Anish','A',4,'HARI','A',9845612355,'AVB svhool',520525,'DHANKUTA')
```

```python
s.student_details()
```
