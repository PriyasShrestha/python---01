```python
class AgeExceptionError(Exception):
    pass
    
```

```python
def checkage(age):
    if age>=18:
        print('you are eligible')
    else:
        raise AgeExceptionError(f'Error: {age} aged person is nor eligible for vote!')
    
```

```python
try:
    age=int(input('Enter your age'))
    checkage(age)
except AgeExceptionError as e:
    print(e)
except ValueError as i:
    print(i)
```

```python
class InsufficientFundError(Exception):
    pass
```

```python
def withdrawn(withdraw_amount):
    balance_amount=100000
    if balance_amount>=withdraw_amount:
        remaining_amount=balance_amount-withdraw_amount
        print(f'balance withdrawn successfull:\n remaining balance: {remaining_amount}')
    else:
        raise InsufficientFundError(f'Error: Insufficient Balance...!!')
    
    
```

```python
try:
    withdraw_amount=int(input('Enter the amount'))
    withdrawn(withdraw_amount)
except InsufficientFundError as e:
    print(e)
except ValueError as i:
    print('Enter the Intiger Number')
    
```
